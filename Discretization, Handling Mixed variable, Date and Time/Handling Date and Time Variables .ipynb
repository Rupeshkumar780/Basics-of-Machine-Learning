<h1 style="color:red">Handling Date and Time Variables </h1>

```python
import pandas as pd 
import numpy as np 
```

```python
date = pd.read_csv('handling_date.csv')
time = pd.read_csv('handling_time.csv')
```

```python
# In Default when you import any dataset, 'date' and 'time' columns are bydefault are 'object'.
# Make sure to change their data type to datetime64.
```

<h2 style="color:blue">Handling Date</h2>

```python
date.head()
```

```python
date.info()    # by default 'date' column has 'object' datatype 
```

```python
# Converting to datetime datatype 
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

<h3 style='color:brown'>Extract year</h3>

```python
date['date_year'] = date['date'].dt.year         # dt -> datetime
```

```python
date.head()
```

<h3 style='color:brown'>Extract Month and Month Name</h3>

```python
date['date_month'] = date['date'].dt.month
date.head(2)
```

```python
date['date_month_name'] = date['date'].dt.month_name()   # extract the name of the month
date.head(2)
```

<h3 style='color:brown'>Extract Day</h3>

```python
date['date_day'] = date['date'].dt.day
date.head(3)
```

```python
# day of week 
# 0 -> Monday, 1 -> Tuesday, ......  ,  6 -> Sunday
date['day_of_week'] = date['date'].dt.dayofweek
date.sample(3)
```

```python
# day of week - name 
date['day_name'] = date['date'].dt.day_name()
date.head(2)
```

```python
# is weekend 
date['is_weekend'] = np.where(date['day_of_week'].isin([5, 6]), 1, 0)           # check if day_of_week is 5, 6 or not
# date['is_weekend'] = np.where(date['day_name'].isin(['Saturday','Sunday']), 1, 0)
date.sample(3)
```

<h3 style='color:brown'>Extract Week of the year</h3>

```python
date.drop(columns=['product_id', 'city_id', 'orders'], inplace=True)  # currently no use of these columns 
```

```python
date['week_num'] = date['date'].dt.isocalendar().week          # ISO calendar is explicit and standard
date.head()
```

```python
# Extract quarter of the year
date['quarter'] = date['date'].dt.quarter       #  1 → Jan–Mar
date.head(2)
```

<h3 style='color:brown'>Extract Time elapsed b/w dates</h3>

```python
# extract today's date 
import datetime

today = datetime.datetime.today()
today
```

```python
today - date['date']       # pandas series
```

```python
# extract days passed
(today - date['date']).dt.days
```

```python
# Month passed
np.round( (today - date['date']) / np.timedelta64(1,'D') / 30, 0)         # round to '0' decimal places
# Directly assumes: 1 month ≈ 30 days
```

<h2 style="color:blue">Handling Time</h2>

```python
time.head()
```

```python
time.info()         # here also 'date' columns has 'object' datatype
```

```python
time['date'] = pd.to_datetime(time['date'])      # Converting to datetime datatype
```

```python
time.info()
```

<h3 style='color:brown'>Extracting Time</h3>

```python
time['hour'] = time['date'].dt.hour 
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

<h3 style='color:brown'>Extracting Time part from datetime</h3>

```python
time['time'] = time['date'].dt.time 
time.head(2)
```

<h3 style='color:brown'>Extracting Time difference</h3>

```python
today - time['date']
```

```python
(today - time['date']) / np.timedelta64(1,'s')    # time difference in seconds
```

```python
(today - time['date']) / np.timedelta64(1,'m')    # time difference in min
```

```python
(today - time['date']) / np.timedelta64(1,'h')    # time difference in hours
```
