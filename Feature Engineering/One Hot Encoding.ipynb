```python
# Train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df['selling_price'], test_size=0.25, random_state=0)
X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
# ohe = OneHotEncoder()    # create a object of OneHotEncoder

# ohe = OneHotEncoder(drop='first')      #  to drop first column
# ohe = OneHotEncoder(drop='first', sparse_output=False)      
#  it will directly convery into numpy array, no need to use .toarray() later on, else wise it will create a sparse matrix
ohe = OneHotEncoder(drop='first', dtype=np.int32)      #  we can control the datatype of the array formed, bydefault float data type is there
```

```python
# apply OneHotEncoder to the columns seperately and then merge them with the remaining columns

# train and transform X_train at once
X_train_new = ohe.fit_transform(X_train[['fuel','owner']]).toarray()       # OneHotEncoder bydefault produces a sparse matrix 

# for X_test we only perform transform (always)
X_test_new = ohe.transform(X_test[['fuel','owner']]).toarray()         # sparse matrix -> toarray()   convert into a numpy array

X_train_new
```

```python
#  now we append "brand" and "km_driven" into the X_train_new

npArray = X_train[['brand', 'km_driven']].values        # data converted to numpy array, 
# we are not applying OneHotEncoding on brands now, as it creates a lot of columns

#  to horizontally stack both numpy arrays npArray and X_Train_new, use np.hstack()
X_train_new = np.hstack((npArray, X_train_new))          # pass both numpy array in a tuple form ()

npArrayTest = X_test[['brand','km_driven']]
X_test_new = np.hstack((npArrayTest, X_test_new))        # all the input array dimensions except for the concatenation axis must match exactly 

X_train_new
```

<h2 style="color:blue">OneHotEncoding with Top Categories</h2>

<h3 style="color:brown">Setting Threshold</h3>

```python
brandCount = df['brand'].value_counts()       # store the (brand name and cars count) series in a variable
brandCount                     # brandCount is a pandas series
```

```python
# we take a threshold of 100. 
# cars count above 100 will have a seperate column while <100 will be clubbed together in 'others' coulmn
threshold = 100
```

```python
# applying logic, take out the brands name whose threshold is less than 100. 
replaceBrands = brandCount[brandCount <= threshold].index
replaceBrands
```

```python
df_updated = df.copy()          # avoid changing original dataframe
df_updated['brand'] = df_updated['brand'].replace(replaceBrands, 'Other_Brands')
display(df_updated.sample(5))
df_updated['brand']          # df_updated['brand'] is a pandas Series
```

<h3 style="color:brown">pandas get_dummies()</h3>

```python
pd.get_dummies(df_updated, columns=['fuel','owner','brand'],dtype=np.int32, drop_first=True)
```

<h3 style="color:brown">sklearn OneHotEncoder</h3>

```python
df_updated.sample(5)
```

```python
# train test split
X_train, X_test, y_train, y_test = train_test_split(df_updated.iloc[:, :-1], df_updated["selling_price"], test_size=0.25, random_state=0)

X_train.shape, X_test.shape
```

```python
X_train.sample(5)
```

```python
# using sklearn library
# updatedBrand.to_frame()        # if updatedBrand -> pandas series, Convert to DataFrame

ohe = OneHotEncoder(drop='first', dtype=np.int32, sparse_output=False)
X_train_encodedCols = ohe.fit_transform(X_train[['brand','fuel','owner']])
X_train_encodedCols.shape
```

```python
# X_train_encodedCols is a numpy array

# ohe.get_feature_names_out()   #  sklearn keeps encoded column names stored in this
encoded_col_names = ohe.get_feature_names_out()  
X_train_encodedCols = pd.DataFrame(X_train_encodedCols, columns=encoded_col_names, index=X_train.index)
# index -> assigns the same row labels as X_train to the encoded DataFrame so pandas can align rows correctly later.
X_train_encodedCols
```

```python
# other_columns = X_train.drop(X_train[['brand','fuel','owner']], axis=1)
# other_columns

# concatenate the rest columns of X_train with X_train_encodedCols 
# axis=1 tells pandas to JOIN DATA SIDE-BY-SIDE (add columns)
X_train_new = pd.concat([X_train['km_driven'], X_train_encodedCols], axis=1)        # Pandas concatenate rows using index values, not row order.
X_train_new
```
