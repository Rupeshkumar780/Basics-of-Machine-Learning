```python
# we gonna add a new passenger details, and use the objects stored in 'models/....' 
# and then we will precict whether the passenger would survive or not
```

```python
import pickle
import numpy as np
```

<h3 style="color:brown">Load saved encoders and trained model</h3>

```python
# importing objects from the 'models/...' with the same name as used while exporting
# these objects will behave same as they where behaving in the original file 

ohe_sex = pickle.load(open('models/ohe_sex.pkl', 'rb')) 
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl','rb'))
dt = pickle.load(open('models/decision_tree.pkl', 'rb'))
```

<h3 style="color:brown">Load saved encoders and trained model</h3>

```python
# Assume user input
# Pclass/gender/age/SibSp/Parch/Fare/Embarked 
test_input = np.array([2,'male', 34, 1,2, 36.2, 'S'], dtype=object).reshape(1,7)
```

```python
test_input
```

<h3 style="color:brown">Apply OHE</h3>

```python
# apply all the preprocesing steps which you earlier applied on the train data,
# like imputer, OHE, then concatenate and then apply decision tree
```

```python
# Loads the trained OneHotEncoder used for the 'Sex' column

# better practice to use column no. like (1 -> sex) instead of using like df['sex'], 
test_input_sex = ohe_sex.transform(test_input[:, 1].reshape(1,1)) 
# This warning occurs because the encoder was trained using a DataFrame column name (Sex), 
# but during prediction you passed a NumPy array without column names.
```

```python
test_input_sex      # OHE encoded for 'male'
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:, -1].reshape(1,1))
```

```python
test_input_embarked      # OHE encoded for 'S'
```

```python
test_input_age = test_input[:, 2].reshape(1,1)
```

```python
test_input_age
```

<h3 style="color:brown">Concatenating all numpy arrays</h3>

```python
test_input_tnf = np.concat((test_input_age, test_input_sex, test_input[:, [0,3,4,5]], test_input_embarked), axis=1)
```

```python
display(test_input_tnf.shape)
test_input_tnf
```

<h3 style="color:brown">Applying Decision tree</h3>

```python
dt.predict(test_input_tnf)    # it predicts that the passenger wouldn't survive
```
