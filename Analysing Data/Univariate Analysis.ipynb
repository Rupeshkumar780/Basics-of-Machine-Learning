<h1> Exploratory Data Analysis (EDA)</h1>

```python
# The preliminary analysis of data to discover relationships between measures in the data and to gain an insight on the 
# trends, patterns, and relationships among various entities present in the data set with the help of 
# statistics and visualization tools is called Exploratory Data Analysis (EDA). 
```

<h2 style="color:red"> Univariate Analysis</h2>

```python
# Uni means one and variate means variable, so in univariate analysis, there is only one dependable variable.
```

```python
import pandas as pd 
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

<h2 style="color:red"> 1. Categorical Data</h2>

<h3 style="color:blue">a). countplot()</h3>

```python
display(df["Pclass"].value_counts())
df["Pclass"].value_counts().plot(kind='bar')
```

```python
import seaborn as sns               #  Advanced statistical visualizations
sns.countplot(x=df['Survived'])     # countplot() expects keyword arguments like x= or y=
```

<h3 style="color:blue">b). PieChart</h3>

```python
df['Embarked'].value_counts().plot(kind='pie', autopct="%.2f")
```

```python
df['Sex'].value_counts().plot(kind='pie', autopct="%.2f")
```

<h2 style="color:red"> 2. Numerical Data</h2>

<h3 style="color:blue">a). HistoGram </h3>

```python
import matplotlib.pyplot as pyplot      #  Basic plotting library
```

```python
pyplot.hist(df['Age'])
pyplot.xlabel("Age")               # X-axis label
pyplot.ylabel("Frequency")         # Y-axis label
pyplot.title("Age Distribution")   # Plot title
pyplot.show()
```

```python
pyplot.hist(df['Age'],bins= 20)           # bins helps you to decide in how many intervals you want to distribute the data
```

```python
sns.histplot(df["Age"],bins= 10, kde=True)   # kde=True adds density curve,    KDE = smooth probability curve
pyplot.title("Age Distribution with KDE")
pyplot.show()
```

<h3 style="color:blue">b). DistPlot </h3>

```python
sns.distplot(df["Age"],bins= 10)
```

```python
# A box plot summarizes a numerical variable using quartiles, median, spread, and outliers in a single visual.
# Dots outside whiskers = outliers,  Median line inside box

sns.boxplot(x=df['Fare'])      # we can remove outliers as these may effect the expected outcome
pyplot.title("Boxplot for Fare")
pyplot.show()
```

```python
sns.boxplot(x = df['Age'])
```
