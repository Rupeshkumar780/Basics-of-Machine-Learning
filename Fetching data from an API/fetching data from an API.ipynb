```python
import pandas as pd
import requests         # python library,it helps python to talk with websites or APIs
```

<h2 style="color:brown">Reading API data (json) </h2>

<p><b>Public Cat API</b></p>

```python
response = requests.get('https://api.thecatapi.com/v1/images/search?limit=10')  
# response : This response contains everything the server sent back, including:
# Status code (success or failure)
# Headers (metadata)
# Actual data (body)

# Status Code
###   response.status_code   ###
# Common status codes:
# 200 ->	Success 
# 400 ->	Bad request
# 401 ->	Unauthorized 
# 404 ->	Not found
# 500 ->	Server error
```

```python
data = response.json()               #  returns data in json format   ## in this case it's a list of python dictionaries(key-value pairs)
print(type(data))                    # prints the type of data -> "list"
print(type(data[0]["url"]))            # prints the type of data["url"] -> str
print(data[0].keys())                   # keys of the dict data[0],  in data -> list of dictionaries,
# display(data)                # it will display data in json format

df = pd.DataFrame(response.json())   #  converting json format into pandas dataframe
print(type(df))
display(df)                          #  displays Entire DataFrame
# display(df.head())                 #  First 5 rows (by default)
# display(df.head(10))               #  First 10 rows
```

<h2 style="color:brown">Using For Loop </h2>

<p><b>OpenWeather forecast API</b></p>

```python
API_KEY = "955976b11d772a8e251a7bc80c033dae"  
CITY = "DELHI"

URL = "https://api.openweathermap.org/data/2.5/forecast"
params = {
    "q": CITY,
    "appid": API_KEY,
    "units": "metric"
}

response = requests.get(URL, params=params)
data = response.json()
print(type(data))            # Print the type of the returned JSON object (should be dict)
print(data.keys())           # Print all top-level keys present in the JSON response

# FETCH DATA
weather_report = []    ## empty python list to store processed weather records
# display(data["list"])

for item in data["list"]:                # Loop through each forecast entry in the 'list' key
    temp = item["main"]["temp"]
    humidity = item["main"]["humidity"]
    desc = item["weather"][0]["description"]
    timestamp = item["dt_txt"]
    # Create a dictionary representing one row of weather data
    records = {"temperature":temp, "humidity":humidity, "description":desc, "timestamp":timestamp}

    weather_report.append(records)

# Convert the list of dictionaries into a Pandas DataFrame   
df4 = pd.DataFrame(weather_report)
display(df4)
```

<p><b>jsonplaceholder API</b></p>

```python
df5 = pd.DataFrame()  ## empty dataFrame

# Loop through pages 1 to 10 (JSONPlaceholder supports pagination)
for page in range(1,11):
    url = "https://jsonplaceholder.typicode.com/posts"
    params = {"page":page, "_limit":100 }

    data = requests.get(url, params=params)
    temp_df = pd.DataFrame(data.json())
    
    # Concatenate the new page DataFrame with the main DataFrame
    # ignore_index=True resets the index so it remains continuous (0, 1, 2, ...)
    df5 = pd.concat([temp_df, df5], ignore_index=True)    

print(df5.shape)
display(df5)


# pandas removed .append() Because this pattern is very inefficient:
# bcz It Copies the entire DataFrame every loop
# Is extremely slow and Wastes memory

# So pandas forced best practices.  -->   replacement:   pd.concat()
```

<h2 style="color:brown">Dataframe with images instead of url</h2>

```python
from IPython.display import HTML, display

url = "https://api.thecatapi.com/v1/images/search?limit=10"
response = requests.get(url)

df = pd.DataFrame(response.json())
display(df)

df["url"] = df["url"].apply(
    lambda x: f'<img src="{x}" width="100"/>'
)

display(HTML(df.to_html(escape=False)))
```
