<h1 style="color:red">Web Scraping</h1>

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup          # web scraping Library
```

<h2 style="color:BLUE">Using BeautifulSoup web Scraping Library</h2>

<h3 style="color:brown">Extracting outer-most div of each company </h3>

```python
companyDetails = []    # empty python list in which we will store all the company details

for i in range(1, 51):
    url = "https://www.ambitionbox.com/list-of-companies"
    params = {"page":i}
    headers = {
        "User-Agent": "Mozilla/5.0",       # Identifies who is making the request
        "Accept": "application/json"       # Tells the server what response format you want
    }
    webpage = requests.get(url, params=params, headers=headers).text
    bsoup = BeautifulSoup(webpage, 'lxml')
    companies = bsoup.find_all('div', class_="companyCardWrapper")

    for comp in companies:
        name = comp.find('h2').text.strip()                                     # only one h2 tag is there, so we are using find() function
        rating = comp.find('div', class_="rating_text").text.strip()            # find returns the first occurrence, where we will get ratings
        try:
            about = comp.find('span', class_="companyCardWrapper__interLinking").text.strip()
        except AttributeError:               #  Catch only expected error (AttributeError)
            about = "NaN"
        try:
            ratingComparison = comp.find('div', class_="companyCardWrapper__ratingComparisonWrapper").text.strip()
        except AttributeError:               #  Catch only expected error (AttributeError)
            ratingComparison = "NaN"
        
        compDetails = comp.find_all('a', class_="companyCardWrapper__ActionWrapper")
        reviews = compDetails[0].text.strip()
        salaries = compDetails[1].text.strip()
        interviews = compDetails[2].text.strip()  
        jobs = compDetails[3].text.strip()  
        benefits = compDetails[4].text.strip()  
        
    
        companyData = {
            "Company Name":name,  "Ratings":rating,
            "Description":about,  "Comparison":ratingComparison,
            "Reviews": reviews,   "Salary": salaries,
            "Interviews":interviews,  "Jobs":jobs,  "Benefits":benefits
        }
    
        companyDetails.append(companyData)


df = pd.DataFrame(companyDetails)
print(df.shape)
display(df)
```

```python
df.to_csv("Company_")
```
